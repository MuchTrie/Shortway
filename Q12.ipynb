```python
import re

XML_FILE = "_sysmon_records.xml"

# Ambil nilai berdasarkan Name
def get(event, name):
    pattern = r'Name="' + re.escape(name) + r'">(.*?)</Data>'
    match = re.search(pattern, event, re.DOTALL)
    return match.group(1).strip() if match else ""

# Baca hasil parsing Sysmon
with open(XML_FILE, "r", encoding="utf-8") as f:
    records = f.read().split("<Event ")

# Process target dari Q10
TARGET_PROCESS = "sndvol.exe"

# Cari DNS Query dari SndVol.exe
dns_events = []

print("=" * 70)
print("Q12 - DNS QUERY DARI SNDVOL.EXE")
print("=" * 70)

for event in records:

    event_id = re.search(
        r"<EventID[^>]*>(\d+)</EventID>",
        event
    )

    if not event_id or event_id.group(1) != "22":
        continue

    image = get(event, "Image")
    query = get(event, "QueryName")

    if not image or not query:
        continue

    if TARGET_PROCESS in image.lower():

        dns_events.append({
            "UtcTime": get(event, "UtcTime"),
            "ProcessId": get(event, "ProcessId"),
            "Image": image,
            "QueryName": query
        })

        print("UtcTime   :", get(event, "UtcTime"))
        print("ProcessId :", get(event, "ProcessId"))
        print("Image     :", image)
        print("QueryName :", query)
        print("-" * 70)

# Ambil domain unik dan urutkan
domains = sorted(
    set(event["QueryName"] for event in dns_events)
)

# Tampilkan domain
print()
print("=" * 70)
print("UNIQUE DOMAINS")
print("=" * 70)

for domain in domains:
    print(domain)

# Tampilkan jawaban
print()
print("=" * 70)
print("FINAL ANSWER Q12")
print("=" * 70)

if len(domains) >= 2:
    print(",".join(domains))
else:
    print("Belum ditemukan minimal dua domain.")
```
