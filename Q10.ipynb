```python
import re

XML_FILE = "_sysmon_records.xml"

def get(event, name):
    pattern = r'Name="' + re.escape(name) + r'">(.*?)</Data>'
    match = re.search(pattern, event, re.DOTALL)
    return match.group(1).strip() if match else ""

with open(XML_FILE, "r", encoding="utf-8") as f:
    records = f.read().split("<Event ")

print("=" * 80)
print("SYSMON EVENT ID 8 - CreateRemoteThread")
print("=" * 80)

for x in records:

    event_id = re.search(
        r"<EventID[^>]*>(\d+)</EventID>",
        x
    )

    if not event_id or event_id.group(1) != "8":
        continue

    source = get(x, "SourceImage")
    target = get(x, "TargetImage")

    if source and target:
        print("UtcTime     :", get(x, "UtcTime"))
        print("SourceImage :", source)
        print("TargetImage :", target)
        print("-" * 80)
```
